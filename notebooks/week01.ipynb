## Lesson B: Lists and `for` Loops

A **list** stores a sequence of values:

```python
numbers = [1, 2, 3]
```

A **for loop** lets you go through each value in a list:

```python
for n in numbers:
    print(n)
```

To build a total, create an accumulator variable and update it each loop iteration.

````
# Example: summing a list with a for loop

example_numbers = [4, 5, 6]
example_total = 0

for n in example_numbers:
    example_total = example_total + n

print(example_total)  # 15
````

## Lesson C: Conditionals

A **conditional** lets your code make decisions based on a rule.

```python
if x > 10:
    print("big")
else:
    print("small")
```

You can also chain multiple conditions using `elif`:

```python
if score < 60:
    grade = "F"
elif score < 70:
    grade = "D"
else:
    grade = "C or above"
```

````
# Example: conditionals

temperature = 82

if temperature < 60:
    weather_label = "cold"
elif temperature <= 80:
    weather_label = "warm"
else:
    weather_label = "hot"

print(weather_label)  # hot
````

## Lesson D: Functions

A **function** is a reusable block of code. You define it once and call it many times.

```python
def greet(name):
    return "Hello, " + name
```

Functions help you organize logic and avoid repeating the same code.

````
# Example: function definition and function call

def square(n):
    return n * n

print(square(3))  # 9
print(square(5))  # 25
````
