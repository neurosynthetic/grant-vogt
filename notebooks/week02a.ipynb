## Quick Reminder: What is a function?

A **function** is a block of code that you define once and then call many times.

Functions help you:
- avoid repeating code
- break larger problems into smaller parts
- test your logic one step at a time

Basic pattern:

```python
def function_name(input_value):
    # do something
    return output_value
```


````python
# Example function: define once, call many times

def double(n):
    return 2 * n

print(double(3))
print(double(10))
print(double(-4))
````
